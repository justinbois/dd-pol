```python
import numpy as np

import bokeh.io
import bokeh.plotting

bokeh.io.output_notebook()
```

```python
rng = np.random.default_rng()
```

```python
mu_m = 10 ** rng.normal(0.5, 0.75)
mu_l = 10 ** rng.normal(-2.5, 0.75)
sigma_m_0 = 10 ** rng.normal(-1, 0)
sigma_l_0 = 10 ** rng.normal(-1, 0)
sigma_m = sigma_m_0 * mu_m
sigma_l = sigma_l_0 * mu_l
rho = 2 * rng.beta(20, 6.5) - 1

Sigma = np.array([[sigma_m**2,               rho * sigma_m * sigma_l],
                  [rho * sigma_m * sigma_l,  sigma_l**2]]
                )
mu = np.array([mu_m, mu_l])

rng.multivariate_normal(mu, Sigma, size=60)
```
