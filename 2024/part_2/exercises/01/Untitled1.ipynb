```python
sm = cmdstanpy.CmdStanModel(stan_file='smfish.stan')
```

```python
df = pl.read_csv('../data/singer_transcript_counts.csv', comment_prefix='#')
```

```python
data = dict(N=len(df), n=df['Rest'].to_numpy())
```

```python
data
```

```python
samples = sm.sample(data=data)
samples = az.from_cmdstanpy(posterior=samples)
```

```python
samples
```

```python
p = bokeh.plotting.figure(
    frame_width=250,
    frame_height=250,
    x_axis_label='α',
    y_axis_label='b',
)

p.scatter(
    samples.posterior['alpha'].values.flatten(),
    samples.posterior['b'].values.flatten(),
    size=2,
    alpha=0.2
)

bokeh.io.show(p)
```
